## Exploratory Data Analysis (EDA)

### Univariate Analysis

#### Attrition Distribution

```python
df['Attrition'].value_counts()
```

```python
sns.countplot(x = 'Attrition', data = df ,hue = 'Attrition',palette=['skyblue', 'orange'])
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition ----------------->')
plt.ylabel('Employee Count ------------------>')
plt.show()
```

#### Gender Distribution

```python
df['Gender'].value_counts()
```

```python
sns.histplot(data = df, x = 'Gender', hue = 'Gender',palette = ['skyblue','pink'])
plt.title('Gender Histrogram')
plt.show()
```

#### Department Distribution

```python
x = df['Department'].value_counts()
x
```

```python
plt.figure(figsize=(5,5))
plt.pie(x, labels = x.index, autopct = '%1.1f%%', colors = ['skyblue','lightgreen','orange'], shadow = True, explode = [0.02, 0.02, 0.02],    wedgeprops={'width':0.4}
)
plt.title('Department Distribution')
plt.show()
```

#### MonthlyIncome Distribution

```python
x = df['MonthlyIncome'].value_counts()
x
```

```python
sns.boxplot(df['MonthlyIncome'] , color = 'orange')
plt.title('Distribution of Monthly Income ')
plt.xlabel('Monthly Income ------------>')
plt.show()
```

#### Total working Years Distribution

```python
x = df['TotalWorkingYears'].value_counts()
x
```

```python
sns.kdeplot(df['TotalWorkingYears'], fill = True, color = 'green' )
plt.title('Density Distribution of Total working Years')
plt.xlabel('Total Working Years --------->')
plt.ylabel('Density ----------->')
plt.show()
```

### Bivariate Analysis

```python
cat_cols = df.describe(include = 'O').columns
print('Categorical Columns :- ', cat_cols)
```

#### Attrition vs Gender

```python
sns.countplot(data = df, x = 'Gender', hue = 'Attrition', palette = 'pastel')
plt.title('Gender vs Attrition')
plt.show()
```

#### Attrition vs JobRole 

```python
sns.countplot(data = df, x = 'JobRole', hue = 'Attrition', palette = 'Set3')
plt.title('JobRole vs Attrition')
plt.xlabel('Job Role ----------------->')
plt.ylabel('Employee Count ------------->')
plt.xticks(rotation = 45)
plt.show()
```

```python
num_cols = df.describe().columns
print('Numerical Columns :- ', num_cols)
```

#### DistanceFromHome vs Attrition

```python
sns.boxplot(data = df, x = 'Attrition', y = 'DistanceFromHome', palette = 'viridis')
plt.title('Distance From Home vs Attrition')
plt.show()
```

#### PercentSalaryHike Distribution

```python
plt.figure(figsize = (10,5))
sns.kdeplot(data = df, x = 'PercentSalaryHike',hue = 'Attrition', fill = True , color = 'red')
plt.title('PercentSalaryHike vs Attrition')
plt.xlabel('Percentage Salary Hike --------------->')
plt.ylabel('Density')
plt.grid()
plt.show()
```

#### Crosstab Heatmap

```python
x = pd.crosstab(df['Department'], df['Attrition'])
x
```

```python
sns.heatmap(x , annot = True,cmap ='YlGnBu', fmt = 'd')
plt.title('Department vs Attrition Heatmap')
plt.show()
```
